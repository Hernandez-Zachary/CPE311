```python
import pandas as pd
import numpy as np
import requests

data = requests.get('https://data.cityofnewyork.us/resource/2upf-qytp.csv')
```

```python
data.status_code
```

```python
data.text
```

```python
#Since requests code only converts it into raw data, we have to sort it by converting the raw string to file like structure.
from io import StringIO

## Converts raw string into file structure
string_io_object = StringIO(data.text)

## decode encoded string (sep="," separates them into columns per comma (,))
data_df = pd.read_csv(string_io_object, sep=",")

#Create a DataFrame by reading in the 2019_Yellow_Taxi_Trip_Data.csv file. Examine the first 5 rows.
data_df = pd.DataFrame(data_df)
data_df.head()
```

```python
#Find the dimensions (number of rows and number of columns) in the data.
data_df.shape
```

```python
#Isolate the fare_amount, tip_amount, tolls_amount, and total_amount for the longest trip by distance (trip_distance).
# idxmax finds the row of trip_distance with the highest value
longest_trip = data_df['trip_distance'].idxmax()

# value (how long was the distance)
longest = data_df.loc[longest_trip, 'trip_distance']

# The isolated result of the longest trip
isolated = data_df.loc[longest_trip, ['fare_amount', 'tip_amount', 'tolls_amount', 'total_amount']]

# Display the results separately
print(f"Longest Distance: {longest}")
print("Detail:")
print(isolated)
```
