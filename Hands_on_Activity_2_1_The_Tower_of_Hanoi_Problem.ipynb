```python
#Function to display the Towers of Hanoi
def Display(A,B,C):
    print("Source Peg:     ", A)
    print("Auxiliary Peg:  ", B)
    print("Destination Peg:", C)
    print("-" * 20)

#Initialization of peg lists
print("Initial States of pegs")
A = [50, 15, 10, 5, 1]
B = []
C = []
Display(A, B, C)

n = (len(A))
def TOHLogic(n, A, B, C, A_name='Source Peg', B_name='Auxiliary Peg', C_name='Destination Peg'):
#Phase 1
    if n == 1:
      disk = A.pop()
      C.append(disk)
      print(f"Moving {disk} : {A_name} to {C_name}")
      Display(A, B, C)
      return

    TOHLogic(n-1, A, C, B, A_name, C_name, B_name)

    disk = A.pop()
    C.append(disk)
    print(f"Moving {disk} : {A_name} to {C_name}")
    Display(A, B, C)

  #Phase 2
    TOHLogic(n-1, B, A, C, B_name, A_name, C_name)

n = len(A)
TOHLogic(n, A, B, C)



```
